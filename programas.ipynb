```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

texto = "Estar em toda a parte onde houver bem a fazer e sofrimento a aliviar"

# Conjunto de palavras que vão ser removidas do texto

as_minhas_stopwords = set(['em' , 'toda' , 'a' , 'onde' , 'e'])

# ------------------------- Configurar e gerar a nuvem a partir do texto -----------------------------

nuvem = WordCloud (
    width=800,
    height=400,
    background_color='#fcbf49',
    colormap='viridis',
    max_words=6, # nº máximo e palavras no cartaz
    stopwords=as_minhas_stopwords #remover as minhas palavras
).generate(texto)

# ----------------------- Exibir a imagem através da Matplotlib (biblioteca python usada para criar gráficos) -------------------------

plt.figure(figsize=(10, 5)) # Tamanho da exibição da figura
plt.imshow(nuvem, interpolation = 'bilinear') # Renderiza a imagem
plt.axis('off')
plt.show

nuvem.to_file('nuvem_palavras.png')
```

```python
resolucao = int(input('Introduza a resolução da imagem em pixels'))

if resolucao < 0:
    print ('Por favor introduza uma resolução válida')

if resolucao < 1200:
    categoria = 'Resolução Standard - SD'

elif 1280 < resolucao <= 1919:
    categoria = 'Resolução HD'

elif 1920 < resolucao <= 3840:
    categoria = 'Resolução FHD (2K)'

else:
    categoria = 'Resolução Ultra HD (4K)'

print (f'Categoria: {categoria}')
```

```python
import random

print ('-------------------------------------- BEM VINDO AO JOGO DO NÚMERO SECRETO -----------------------------------------------')

numero_secreto = random.randrange (1, 101)
total_de_tentativas = 0

print ('Qual o nível de dificuldade?')
print ('1 - Facil, 2 - Intermédio, 3 -  Difícil')
nivel = int(input('Digita o teu nível:'))

if nivel == 1:
    total_de_tentativas = 30
elif nivel == 2:
    total_de_tentativas = 20
elif nivel == 3:
    total_de_tentativas = 10

for jogada in range(1, total_de_tentativas + 1):

    print (f'Tentativa {jogada} de {total_de_tentativas}')
    numero_escolhido = input ('Digite um número entre 1 e 101: ')
    print ('O valor escolhido foi:' , numero_escolhido)
    
    # CORREÇÃO AQUI (adicionado o 'c' em escolhid_o_)
    valor_escolhido = int(numero_escolhido)

    if valor_escolhido < 1 or valor_escolhido > 101:
        print ('Número Inválido. O número deve ser entre 1 e 101')
        continue
    
    acertou = valor_escolhido == numero_secreto
    maior = valor_escolhido > numero_secreto
    menor = valor_escolhido < numero_secreto

    if acertou:
        print ('Parabéns! Acertaste o número secreto.')
        break # Para o jogo parar
    else: # pistas
        if maior:
            print ('Ainda não foi desta! O número escolhido foi maior do que o número secreto.')
        elif menor:
            print ('Ainda não foi desta! O número escolhido foi menor do que o número secreto.')

print ('Fim de Jogo')
```
